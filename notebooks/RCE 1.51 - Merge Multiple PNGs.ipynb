```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
def plot_images_in_grid(image_paths, grid_dim, output_path=None):
    fig, axes = plt.subplots(
        nrows=grid_dim[0], ncols=grid_dim[1], figsize=(14, 10)
    )

    for idx, ax in enumerate(axes.flatten()):
        if idx < len(image_paths):
            img = Image.open(image_paths[idx])
            img = np.array(img)
            ax.imshow(img)
            ax.axis("off")
        else:
            ax.axis("off")

    plt.subplots_adjust(
        wspace=0.0, hspace=0.0
    )  # Adjust spacing between images
    if output_path:
        plt.savefig(output_path, dpi=300, bbox_inches="tight")
    else:
        plt.show()
```

```python
BASE_DIR = "/gws/ssde/j25a/ai4er/users/pn341/climate-rl"
IMGS_DIR = f"{BASE_DIR}/results/imgs"
```

```python
exp_ids = [
    "rce-v0-optim-L-10k",
    "rce17-v0-optim-L-10k",
    "rce17-v1-optim-L-10k",
]

stub = "log10_top3_episodic_returns"

img_paths = [IMGS_DIR + "/png/" + x + f"_{stub}.png" for x in exp_ids]
```

```python
output_path = f"{IMGS_DIR}/merged/rce17_x3_v01_merged_{stub}.png"

plot_images_in_grid(img_paths, (3, 1), output_path)
```
